```python
import pandapower as pp
net = pp.create_empty_network()
```

```python
b_hv = pp.create_bus(net, vn_kv=138, name="HV Bus")
b_mv = pp.create_bus(net, vn_kv=12.47, name="MV Bus")
b_load = pp.create_bus(net, vn_kv=12.47, name="Load Bus")
```

```python
pp.create_ext_grid(net, bus=b_hv, vm_pu=1.02, name="Utility Source")
```

```python
pp.create_transformer_from_parameters(net, hv_bus=b_hv, lv_bus=b_mv, 
                                      sn_mva=25, vn_hv_kv=138, vn_lv_kv=12.47,
                                      pfe_kw=30, vkr_percent=0.5, vk_percent=8,
                                      i0_percent=0.1,name="Transformer")
```

```python
pp.create_line_from_parameters(net, from_bus=b_mv, to_bus=b_load, length_km=3,
                               r_ohm_per_km=0.3, x_ohm_per_km=0.35, c_nf_per_km=10,
                               max_i_ka=0.4, name="Line")
```

```python
pp.create_load(net, bus=b_load, p_mw=8, q_mvar=3, name="Load")
```

```python
print(net)
```

```python
pp.runpp(net)
net.res_bus
```
